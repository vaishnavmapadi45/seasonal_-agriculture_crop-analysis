```python
df = df.drop_duplicates().copy()
print('Duplicates remaining', df.duplicated().sum())
```

```python
# Pre-compute the counts used in the dashboard
state_counts = df['State'].value_counts()
crop_counts = df['Crop'].value_counts()

fig, axes = plt.subplots(2, 2, figsize=(18, 12))
fig.suptitle('Seasonal Agriculture Performance Dashboard', fontsize=20)

# Plot 1: Farms by State
state_counts.plot(kind='bar', ax=axes[0, 0])
axes[0, 0].set_title('1. Number of Farms by State', fontsize=14)
axes[0, 0].set_xlabel('State', fontsize=12)
axes[0, 0].set_ylabel('Number of Farms', fontsize=12)
axes[0, 0].tick_params(axis='x', rotation=45)

# Plot 2: Farms by Crop
crop_counts.plot(kind='bar', ax=axes[0, 1])
axes[0, 1].set_title('2. Number of Farms by Crop', fontsize=14)
axes[0, 1].set_xlabel('Crop', fontsize=12)
axes[0, 1].set_ylabel('Number of Farms', fontsize=12)
axes[0, 1].tick_params(axis='x', rotation=45)

# Plot 3: Profit Distribution by Crop
sns.boxplot(x='Crop', y='Profit_INR', data=df, ax=axes[1, 0])
axes[1, 0].set_title('3. Profit Distribution by Crop', fontsize=14)
axes[1, 0].set_xlabel('Crop', fontsize=12)
axes[1, 0].set_ylabel('Profit (INR)', fontsize=12)
axes[1, 0].tick_params(axis='x', rotation=45)

# Plot 4: Profit Distribution by Season
sns.boxplot(x='Season', y='Profit_INR', data=df, ax=axes[1, 1])
axes[1, 1].set_title('4. Profit Distribution by Season', fontsize=14)
axes[1, 1].set_xlabel('Season', fontsize=12)
axes[1, 1].set_ylabel('Profit (INR)', fontsize=12)
axes[1, 1].tick_params(axis='x', rotation=45)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])  # Adjust layout to prevent title overlap
plt.show()
```
